```python
import ast

import pandas as pd
from tornado import options

movie=pd.read_csv("D:/tmdb_5000_movies.csv")
credit=pd.read_csv("tmdb_5000_credits.csv")
movie.head(1)
credit.head(1)
```

```python
movie=movie.merge(credit,on="title")
# movie.head(1)
movie.info()
```

```python
#genres, id, title,keywords, original_language, crew,cast
movie=movie[['genres','id', 'title','keywords','overview','crew','cast']]
```

```python
def convert(obj):
    l = []
    for i in ast.literal_eval(obj):
        l.append(i["name"])
    return l
```

```python
movie["genres"].apply(convert)
movie["genres"]=movie['genres'].apply(convert)
```

```python
movie["keywords"]=movie["keywords"].apply(convert)
```

```python
def convert3(obj):
    l=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter !=3:
            l.append(i["name"])
            counter+=1
        else:
            break
    return l
```

```python
movie['cast']=movie['cast'].apply(convert3)
```

```python
movie.head()
```

```python
def fetch_director(obj):
    l=[]
    for i in ast.literal_eval(obj):
        if i['job']=="Director":
            l.append(i['name'])
            break
    return l
```

```python
# movie["crew"].apply(fetch_director)
movie["crew"]=movie["crew"].apply(fetch_director)
```

```python
movie.head()
```

```python
movie['overview']=movie['overview'].fillna("")
# movie['overview'].isnull().sum()
movie['overview']=movie["overview"].apply(lambda x:x.split())


movie["id"] = movie["id"].astype(str).apply(lambda x: [x])
```

```python
movie.head()
```

```python
movie['genres']=movie['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movie['keywords']=movie['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movie['crew']=movie['crew'].apply(lambda x:[i.replace(" ","") for i in x])
movie['cast']=movie['cast'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movie.head()
```

```python
movie['tags']=(movie["genres"]+movie["keywords"]+movie["overview"]+movie["crew"]+movie["cast"])

# movie["tags"] = (
#     movie["genres"]
#     + movie["keywords"]
#     + movie["overview"]
#     + movie["cast"]
#     + movie["crew"]
#     + movie["id"]
# )
```

```python
new_df=movie[["id","title","tags"]]
```

```python
new_df
```

```python
new_df["tags"][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words="english")
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'][0]
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
sorted(
    [(i, float(score)) for i, score in enumerate(similarity[0])],
    reverse=True,
    key=lambda x: x[1]
)[1:6]
```

```python
def recommend(movie):
    movie_index=new_df[new_df['title']==movie].index[0]
    distance=similarity[movie_index]
    movie_list=sorted([(i, float(score)) for i, score in enumerate(distance)],reverse=True,key=lambda x: x[1])[1:6]

    for i in movie_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Batman Begins")
```
